```python
#Center of NGC3982 in pixels 111,129
#Drawing the apperture

%matplotlib ipympl
flux_list=[]
area_list=[]
plt.figure(figsize=(5,5))
from photutils.aperture import CircularAperture, ApertureStats
plt.imshow(NGC3982crop,norm=LogNorm(),origin='lower',cmap='jet')
rad=np.array(linspace(1,55,100))
for i in range(len(rad)):
  aper3=CircularAperture([129,111],rad[i])
  aper3.plot(color='red',lw=2)

  #find flux in the corresponding aperture
  #phot_table=aperture_photometry(NGC3982crop,aper3)
  #print(i,phot_table)
  aperstats= ApertureStats(NGC3982crop, aper3)
  area_list.append(aperstats.sum_aper_area)
  flux_list.append(aperstats.sum)
  print(aperstats.sum,aperstats.sum_aper_area)
```

```python
print(flux_list)
np.shape(flux_list)
```

```python
plt.figure()
plt.scatter(rad,log(flux_list), marker='o', s=10)
plt.plot(rad,log(flux_list))
plt.title('Curve growth')
plt.xlabel('Radius[Apperture]')
plt.ylabel('Flux value')
plt.show()
```

```python
plt.figure()
#Scale 0.27 arcsec per pixel MuSCAT3 2m Faulkes Telescope North
rad_arc=0.27*rad
plt.scatter(rad_arc,log(flux_list), marker='o', s=10)
plt.plot(rad_arc,log(flux_list))
plt.title('Curve of growth')
plt.xlabel('Radius[Arcsec]')
plt.ylabel('Flux value')
plt.show()
```

$\text{Flux density(i+1 ,i)}=flux/{\pi(r^2_{i+1}-r^2_{i})}$

```python
radiussq=rad_arc**2
raddiff=np.diff(radiussq)
diff_r=np.append(rad_arc[0]**2,raddiff)

flux_diff=np.diff(flux_list)
flux_diff1=append(flux_list[0],flux_diff)
flux_density=flux_diff1/(np.pi * diff_r)
print((flux_density))

plt.figure()
plt.scatter(rad_arc,flux_density,s=15)
plt.loglog(rad_arc,(flux_density))
plt.title('Radial Flux Profile')
plt.xlabel('Radius[Arcsec]')
plt.ylabel('Flux Density')
plt.show()
```

```python
#conversion of flux into magnitudes
mag=22.5-np.log10(flux_density)
#Conversion into arcsec
mag_arcsec=22.5-np.log10(flux_density/(0.27)**2)
```

Magnitude is inversely proportional to Flux density.

```python
#Surface brightness profile

plt.figure()
plt.scatter(rad_arc, mag_arcsec)
plt.plot(rad_arc, mag_arcsec)
plt.title('Surface Brightness Profile [NGC3982]')
plt.xlabel('Radius[Arcsec]')
plt.ylabel('Magnitude/arcsec^2')
plt.show()

plt.figure(figsize=(8,5))
plt.loglog(rad_arc, mag_arcsec, color='r')
plt.title('Magnitude across radius [NGC3982]')
plt.xlabel('Log Radius')
plt.ylabel('Log Magnitude')
plt.show()
```

```python
from astropy.modeling.models import Sersic1D
from scipy.optimize import curve_fit
```

```python
plt.figure()
plt.subplot(111, xscale='log', yscale='log')
s1 = Sersic1D(amplitude=1, r_eff=5)
r=np.arange(0, 100, .01)

for n in range(1, 10):
     s1.n = n
     plt.plot(r, s1(r), color=str(float(n) / 15))

plt.axis([1e-1, 30, 1e-2, 1e3])
plt.xlabel('log Radius')
plt.ylabel('log Surface Brightness')
plt.text(.25, 1.5, 'n=1')
plt.text(.25, 300, 'n=10')
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
########### DEF SERSIC FN ###########
# Define Sersic function as a function
def sersic_single(R, i_b, r_b, n_b):
        return Sersic1D(i_b, r_b, n_b) (R)


########### Single SERSIC FIT ###########
# Fit a single Sersic profile to our data
#I_b = 800
#R_b = 180
#n_b = 2.9
#guess_1=sersic_single(rad_arc, I_b, R_b,n_b)

optgal, covgal = curve_fit(sersic_single,rad_arc,flux_density, bounds=((0,rad_arc[0],0),(1e5,rad_arc[len(rad_arc)-1],10)), maxfev = 10000)
print('Intensity : ', optgal[0])
print('Radius : ', optgal[1])
print('Sersic index : ', optgal[2])
plt.figure()
plt.scatter(rad_arc,flux_density,s=15, label='Data')
plt.loglog(rad_arc,(flux_density), label='Data')
#plt.loglog(rad_arc,guess_1, label='Guess Values')
plt.loglog(rad_arc,sersic_single(rad_arc,*optgal),label='Curve fit',color='r')
plt.title('Radial Flux Profile [NGC3982]')
plt.xlabel('Radius')
plt.ylabel('Flux Density')
plt.legend()
plt.show()
```

```python
# Define Sersic function as a function
def sersic_double(R, i_b, r_b, n_b, i_d, r_d, n_d):
        return Sersic1D(i_b, r_b, n_b) (R) + Sersic1D(i_d, r_d, n_d) (R)


optgal3, covgal3 = curve_fit(sersic_double, rad_arc, flux_density, bounds=((0,rad_arc[0],0,0,rad_arc[0],0),(1e5,rad_arc[len(rad_arc)-1],10,1e5,rad_arc[len(rad_arc)-1],10)), maxfev=10000)


print('Intensity [Bulge]: ', optgal3[0])
print('Radius [Bulge]: ', optgal3[1])
print('Sersic index [Bulge]: ', optgal3[2])

print('Intensity [Disk]: ', optgal3[3])
print('Radius [Disk]: ', optgal3[4])
print('Sersic index [Disk]: ', optgal3[5])


# Define a function for the best fit curve
def galaxyprofile_double(r):
        return Sersic1D(*optgal3[0:3])(r) + Sersic1D(*optgal3[3:6])(r)

# Define a function for the best fit curve
def galaxyprofile_bulge(r):
        return Sersic1D(*optgal3[0:3])(r)

# Define a function for the best fit curve
def galaxyprofile_disc(r):
        return Sersic1D(*optgal3[3:6])(r)


# Make our final Plot
plt.figure()

plt.scatter((rad_arc),(flux_density), label='Data')
plt.loglog(rad_arc,(flux_density))

plt.loglog(rad_arc,galaxyprofile_double(rad_arc) , label='Double fit', color='r')
plt.loglog(rad_arc,sersic_single(rad_arc,*optgal),label='Single fit',color='#FF5732')
plt.plot(rad_arc, galaxyprofile_disc(rad_arc), label='Disc', color='g')
plt.plot(rad_arc, galaxyprofile_bulge(rad_arc), label='Bulge', color='c')


plt.ylim([1e4, 150000])
plt.title('NGC3982 Sersic Model')
plt.xlabel('log Radius')
plt.ylabel('log Surface Brightness')
plt.legend()
plt.show()
```
